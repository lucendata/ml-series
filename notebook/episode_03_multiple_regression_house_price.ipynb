# 🏡 Multiple Regression: Predicting House Price

This notebook walks through building a multiple regression model to predict house prices using `size_sqft` and `num_bedrooms`. We'll use pandas and statsmodels.

```python

import pandas as pd
import statsmodels.api as sm
```

### Step 1: Prepare the data

```python

data = pd.DataFrame({
    'price': [200000, 250000, 300000, 350000, 400000],
    'size_sqft': [1000, 1500, 2000, 2500, 3000],
    'num_bedrooms': [2, 3, 3, 4, 4]
})
data
```

### Step 2: Define predictors and response

#### We add a constant (intercept) to the predictors.
#### This tells the model to calculate the base price when size and bedrooms are zero.


```python

X = data[['size_sqft', 'num_bedrooms']]
X = sm.add_constant(X)
y = data['price']
```

### Step 3: Fit the regression model

#### 💡 Where’s the error term (epsilon) in the code?

In the regression formula:
Y = β₀ + β₁X₁ + β₂X₂ + ε

The ε (epsilon) is the error term — the leftover difference between the actual value and what the model predicts.

In this notebook, we don’t add ε directly in the code.

Instead, when we fit the model:
model = sm.OLS(y, X).fit()

statsmodels automatically calculates **residuals** (actual - predicted), which are our estimates of the error.

For now, we’ll let the library handle the math magic — in later episode, we’ll learn how to check if those residuals look healthy!

```python

model = sm.OLS(y, X).fit()
```

### Step 4: Check the summary

## 📊 Understanding the Regression Output

When you run `model.summary()`, you'll see a detailed table. Here's how to interpret the key components:

- **R-squared**: Indicates how well the model explains the variability of the target variable. A value closer to 1 suggests a better fit.

- **Adjusted R-squared**: Adjusts the R-squared value based on the number of predictors in the model. Useful for comparing models with different numbers of predictors.

- **Coefficients (coef)**: Represent the estimated impact of each predictor on the target variable. For example, a coefficient of 100 for `size_sqft` means that each additional square foot increases the price by \$100.

- **P>|t| (p-value)**: Tests the null hypothesis that the coefficient is zero (no effect). A small p-value (typically ≤ 0.05) indicates that the predictor is statistically significant.

- **F-statistic and Prob (F-statistic)**: Assess the overall significance of the model. A significant F-statistic (with a small p-value) suggests that the model provides a better fit than one with no predictors.

Focus on these elements to understand your model's performance and the influence of each predictor.


```python

print(model.summary())
```

### Step 5: Make predictions for a new house

## 🏠 Making Predictions with the Model

Now that we've trained our model, let's use it to predict the price of a new house. We'll input the size and number of bedrooms, and the model will estimate the price based on the learned relationships.

Feel free to modify the input values to see how changes in size and number of bedrooms affect the predicted price.


```python

new_house = pd.DataFrame({'const': [1], 'size_sqft': [2200], 'num_bedrooms': [3]})
predicted_price = model.predict(new_house)
predicted_price
```
