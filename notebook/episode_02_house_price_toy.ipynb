# Episode 2 – House Price Prediction (Toy Dataset)

In this notebook, we:
- Create a simple dataset of house sizes and prices
- Train a linear regression model
- Predict prices
- Evaluate performance
- Visualize the results

## 1. Create Toy Dataset

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {'Size (sqft)': [500, 750, 1000, 1250, 1500],
        'Price ($)': [150000, 200000, 250000, 275000, 300000]}
df = pd.DataFrame(data)

X = df[['Size (sqft)']]
y = df['Price ($)']

model = LinearRegression()
model.fit(X, y)
```

## 2. View Model Parameters

```python
print("Slope (coefficient):", model.coef_)
print("Intercept:", model.intercept_)
```

## 3. Make Predictions and Evaluate

```python
y_pred = model.predict(X)

from sklearn.metrics import mean_absolute_error, r2_score

mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Predicted prices:", y_pred)
print(f"Mean Absolute Error: ${mae:.2f}")
print(f"R² Score: {r2:.2f}")
```

## 4. Visualize Results

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='blue', label='Actual')
plt.plot(X, y_pred, color='red', label='Prediction')
plt.xlabel("Size (sqft)")
plt.ylabel("Price ($)")
plt.legend()
plt.title("Toy House Price Dataset: Size vs Price")
plt.show()
```

## 5. Summary
- Trained a simple linear regression model
- Predicted house prices from size
- Evaluated performance with MAE and R²
- Visualized predictions against actual data

**Note:** This is a toy dataset to demonstrate the basics of linear regression.
