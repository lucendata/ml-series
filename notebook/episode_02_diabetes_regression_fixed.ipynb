# Episode 2 – Linear Regression with Diabetes Dataset

In this notebook, we:
- Load the diabetes dataset from sklearn
- Train a linear regression model
- Predict disease progression
- Evaluate model performance
- Visualize the results

## 1. Load Diabetes Dataset

```python
from sklearn.datasets import load_diabetes
import pandas as pd

diabetes = load_diabetes(as_frame=True)
df = diabetes.frame

print(df.head())
```

## 2. Train Linear Regression Model (BMI → Disease Progression)

```python
from sklearn.linear_model import LinearRegression

# Use BMI as predictor
X = df[['bmi']]
y = df['target']

model = LinearRegression()
model.fit(X, y)

print("Slope (coefficient):", model.coef_)
print("Intercept:", model.intercept_)

# Make predictions
y_pred = model.predict(X)
```

## 3. Evaluate Model Performance

```python
from sklearn.metrics import mean_absolute_error, r2_score

mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Mean Absolute Error: {mae:.2f}")
print(f"R² Score: {r2:.2f}")
```

## 4. Visualize Predictions

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='blue', alpha=0.5, label='Actual')
plt.plot(X, y_pred, color='red', label='Prediction')
plt.xlabel("BMI")
plt.ylabel("Disease Progression")
plt.legend()
plt.title("Diabetes Dataset: BMI vs Disease Progression")
plt.show()
```

**Note on BMI in the diabetes dataset:**
The BMI numbers you’ll see in the notebook (like 0.03, -0.04) are *standardized* values.  
That means they’ve been preprocessed to have an average near zero, rather than showing raw BMI numbers like 22 or 30.

We’re using this version because it’s built into the dataset — and it still lets us see how BMI affects disease progression.


## 5. Summary
- We trained a linear regression model on the diabetes dataset
- We evaluated the model with MAE and R²
- We visualized the predictions against actual data
